## 6. Download Rendered Video

```python
import os
from google.colab import files

output_file = "out/ChokepointEscalationMaster.mp4"

if os.path.exists(output_file):
    file_size_mb = os.path.getsize(output_file) / (1024 * 1024)
    print(f"✅ Render finished! File size: {file_size_mb:.2f} MB")
    print("⬇️ Initiating download to your computer...")
    files.download(output_file)
else:
    print("❌ Rendered output file not found. Check render logs above for errors.")
```
